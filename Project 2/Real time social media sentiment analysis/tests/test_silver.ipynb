```python
# ==========================================================
# Project      : Social Media Sentiment Analysis
# Notebook     : test_silver
# Description  : Silver Layer Testing
# ==========================================================

tables = [

    "silver_sentiment",
    "silver_trends",
    "silver_tweets",
    "silver_user_metadata",
    "silver_valid_tweets"

]

# ==========================================================
# Test 1 : Silver Tables Exist
# ==========================================================

for table in tables:

    assert spark.catalog.tableExists(
        f"silver_catalog1.processed.{table}"
    ), f"{table} does not exist"

print("✅ Test 1 Passed : All Silver tables exist")


# ==========================================================
# Test 2 : Row Count Validation
# ==========================================================

for table in tables:

    row_count = spark.table(
        f"silver_catalog1.processed.{table}"
    ).count()

    assert row_count > 0, f"{table} is empty"

print("✅ Test 2 Passed : All Silver tables contain data")


# ==========================================================
# Test 3 : silver_tweets Validation
# ==========================================================

required_columns = [

    "tweet_id",
    "tweet_text"

]

df = spark.table(
    "silver_catalog1.processed.silver_tweets"
)

for column in required_columns:

    assert column in df.columns, f"{column} missing"

print("✅ Test 3 Passed")


# ==========================================================
# Test 4 : Null Analysis
# ==========================================================

from pyspark.sql.functions import col

critical_columns = [

    "tweet_id",
    "tweet_text"

]

for column in critical_columns:

    null_count = (
        df.filter(col(column).isNull())
          .count()
    )

    print(f"{column} -> NULL Count : {null_count}")

print("✅ Test 4 Passed")


# ==========================================================
# Test 5 : Duplicate Analysis
# ==========================================================

duplicate_count = (

    df.groupBy("tweet_id")
      .count()
      .filter(col("count") > 1)
      .count()

)

print(f"Duplicate tweet_id records : {duplicate_count}")

print("✅ Test 5 Passed")
```

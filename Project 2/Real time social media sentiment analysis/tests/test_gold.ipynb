```python
# ==========================================================
# Project      : Social Media Sentiment Analysis
# Notebook     : test_gold
# Description  : Gold Layer Testing
# ==========================================================

tables = [

    "fact_social_media",
    "dim_user",
    "dim_topic",
    "dim_date",
    "dim_country",
    "dim_sentiment",
    "country_summary",
    "topic_summary",
    "daily_socialmedia",
    "kpi_dashboard",
    "executive_dashboard"

]

# ==========================================================
# Test 1 : Gold Tables Exist
# ==========================================================

for table in tables:

    assert spark.catalog.tableExists(
        f"silver_catalog1.gold.{table}"
    ), f"{table} does not exist"

print("✅ Test 1 Passed : All Gold tables exist")


# ==========================================================
# Test 2 : Row Count Validation
# ==========================================================

for table in tables:

    row_count = spark.table(
        f"silver_catalog1.gold.{table}"
    ).count()

    assert row_count > 0, f"{table} is empty"

print("✅ Test 2 Passed : All Gold tables contain data")


# ==========================================================
# Test 3 : Fact Table Validation
# ==========================================================

df = spark.table(
    "silver_catalog1.gold.fact_social_media"
)

required_columns = [

    "tweet_id",
    "user_id",
    "topic_key",
    "date_key"

]

for column in required_columns:

    assert column in df.columns, f"{column} missing"

print("✅ Test 3 Passed : Fact table columns validated")


# ==========================================================
# Test 4 : Null Analysis
# ==========================================================

from pyspark.sql.functions import col

for column in ["tweet_id","user_id"]:

    null_count = df.filter(
        col(column).isNull()
    ).count()

    print(f"{column} NULL Count : {null_count}")

print("✅ Test 4 Passed")


# ==========================================================
# Test 5 : Duplicate Tweet Validation
# ==========================================================

duplicate_count = (

    df.groupBy("tweet_id")
      .count()
      .filter(col("count") > 1)
      .count()

)

print(f"Duplicate Tweet IDs : {duplicate_count}")

print("✅ Test 5 Passed")
```

```python
# ==========================================================
# Project      : Social Media Sentiment Analysis
# Notebook     : test_gold
# Description  : Gold Layer Testing
# ==========================================================

spark.sql("USE CATALOG silver_catalog1")
spark.sql("USE SCHEMA gold")

gold_tables = [

    "fact_social_media",
    "dim_user",
    "dim_topic",
    "dim_country",
    "dim_date",
    "dim_sentiment",
    "topic_summary",
    "country_summary",
    "daily_socialmedia",
    "kpi_dashboard",
    "executive_dashboard"

]

# ==========================================================
# Test 1 : Gold Tables Exist
# ==========================================================

for table in gold_tables:

    spark.sql(
        f"SELECT * FROM silver_catalog1.gold.{table} LIMIT 1"
    )

print("✅ Test 1 Passed : All Gold tables exist")


# ==========================================================
# Test 2 : Row Count Validation
# ==========================================================

for table in gold_tables:

    row_count = spark.sql(
        f"SELECT COUNT(*) AS cnt FROM silver_catalog1.gold.{table}"
    ).collect()[0]["cnt"]

    assert row_count > 0, f"{table} is empty"

print("✅ Test 2 Passed : All Gold tables contain data")


# ==========================================================
# Test 3 : Fact Table Validation
# ==========================================================

fact_df = spark.table(
    "silver_catalog1.gold.fact_social_media"
)

required_columns = [

    "tweet_id",
    "user_id",
    "topic_key",
    "date_key",
    "likes",
    "retweets",
    "replies",
    "impressions",
    "engagement_count",
    "sentiment_score"

]

for col_name in required_columns:

    assert col_name in fact_df.columns, \
        f"{col_name} missing in fact_social_media"

print("✅ Test 3 Passed : Fact table columns validated")


# ==========================================================
# Test 4 : KPI Dashboard Validation
# ==========================================================

kpi_df = spark.table(
    "silver_catalog1.gold.kpi_dashboard"
)

assert kpi_df.count() > 0

print("✅ Test 4 Passed : KPI Dashboard contains data")


# ==========================================================
# Test 5 : Executive Dashboard Validation
# ==========================================================

exec_df = spark.table(
    "silver_catalog1.gold.executive_dashboard"
)

assert exec_df.count() > 0

print("✅ Test 5 Passed : Executive Dashboard contains data")


# ==========================================================
# Test 6 : Topic Summary Validation
# ==========================================================

topic_df = spark.table(
    "silver_catalog1.gold.topic_summary"
)

assert topic_df.count() > 0

print("✅ Test 6 Passed : Topic Summary contains data")


# ==========================================================
# Test 7 : Country Summary Validation
# ==========================================================

country_df = spark.table(
    "silver_catalog1.gold.country_summary"
)

assert country_df.count() > 0

print("✅ Test 7 Passed : Country Summary contains data")


# ==========================================================
# Test 8 : Daily Summary Validation
# ==========================================================

daily_df = spark.table(
    "silver_catalog1.gold.daily_socialmedia"
)

assert daily_df.count() > 0

print("✅ Test 8 Passed : Daily Summary contains data")


# ==========================================================
# Test 9 : Null Check on Fact Table
# ==========================================================

from pyspark.sql.functions import col

critical_columns = [

    "tweet_id",
    "user_id",
    "topic_key",
    "date_key"

]

for column in critical_columns:

    null_count = fact_df.filter(
        col(column).isNull()
    ).count()

    print(f"{column} -> NULL Count : {null_count}")

print("✅ Test 9 Passed : Null analysis completed")


# ==========================================================
# Test 10 : Duplicate Tweet Validation
# ==========================================================

duplicate_count = (

    fact_df.groupBy("tweet_id")
           .count()
           .filter("count > 1")
           .count()

)

print(
    f"Duplicate tweet_id records : {duplicate_count}"
)

print("✅ Test 10 Passed : Duplicate analysis completed")


# ==========================================================
# Final Result
# ==========================================================

print("")
print("🎉 ALL GOLD LAYER TESTS PASSED SUCCESSFULLY")
```

```python
#Read Tweets CSV
tweets_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("abfss://src-container@santhoshsocialadls01.dfs.core.windows.net/tweets/bronze_tweets_raw.csv")
```

```python
#Verify Data Loaded

display(tweets_df)
```

```python
#Check Schema
tweets_df.printSchema()
```

```python
spark.sql("USE CATALOG bronze_catalog")
spark.sql("USE SCHEMA social_media")
```

```python
tweets_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze_tweets")
```

```python
spark.sql("SHOW TABLES").show()
```

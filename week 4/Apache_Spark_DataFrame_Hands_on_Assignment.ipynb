```python
# Import SparkSession

from pyspark.sql import SparkSession

# Create Spark Session

spark = SparkSession.builder \
    .appName("EmployeeDataFrameAssignment") \
    .getOrCreate()

spark
```

```python
# Import Required Functions

from pyspark.sql.functions import *
```

```python
# Create Employee Data

data = [
    (101,"John","IT",50000,"Chennai",30),
    (102,"Alice","HR",60000,"Bangalore",28),
    (103,"Bob","Finance",70000,"Mumbai",35),
    (104,"David","IT",80000,"Chennai",40),
    (105,"Eva","HR",55000,"Hyderabad",25),
    (106,"Frank","Finance",90000,"Mumbai",45)
]

# Define Column Names

columns = [
    "emp_id",
    "emp_name",
    "department",
    "salary",
    "city",
    "age"
]
```

```python
# Create Employee DataFrame

emp_df = spark.createDataFrame(
    data,
    columns
)

emp_df.show()
```

```python
# Create Dataset With NULL Salary Values

data_null = [
    (107, "Sam", "IT", None, "Chennai", 29),
    (108, "Tom", "HR", None, "Bangalore", 32)
]
```

```python
# Create DataFrame Using Existing Schema

null_df = spark.createDataFrame(
    data_null,
    schema=emp_df.schema
)

null_df.show()
```

```python
# Combine Original And NULL DataFrames

emp_null_df = emp_df.union(null_df)

emp_null_df.show()
```

```python
# Question 47
# Display Records Having NULL Salaries

emp_null_df.filter(
    col("salary").isNull()
).show()
```

```python
# Question 48
# Replace NULL Salary Values With 50000

emp_null_df.fillna(
    {"salary": 50000}
).show()
```

```python
# Question 49
# Drop Records Containing NULL Values

emp_null_df.na.drop().show()
```

```python
# Question 50
# Count Total NULL Salary Records

print(
    emp_null_df.filter(
        col("salary").isNull()
    ).count()
)
```

```python
# Question 51
# Replace NULL Salary Values Using Department Average Salary

dept_avg = emp_null_df.groupBy(
    "department"
).agg(
    avg("salary").alias("avg_salary")
)

result = emp_null_df.join(
    dept_avg,
    on="department",
    how="left"
).withColumn(
    "salary",
    when(
        col("salary").isNull(),
        col("avg_salary")
    ).otherwise(col("salary"))
).drop("avg_salary")

result.show()
```
